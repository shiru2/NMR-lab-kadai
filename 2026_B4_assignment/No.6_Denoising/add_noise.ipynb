# No.6 ガウスノイズ付加

デノイジング演習の準備として、クリーンな画像にガウスノイズを付加します。

```python
import numpy as np
import skimage.io
import skimage.metrics
import matplotlib.pyplot as plt
import japanize_matplotlib
```

```python
noise_sigma = 0.05  # ノイズ標準偏差

A = skimage.io.imread('data/original.pgm').astype(float) / 255.0

rng = np.random.default_rng(seed=0)
noise = rng.normal(0, noise_sigma, A.shape)
An = np.clip(A + noise, 0, 1)

psnr = skimage.metrics.peak_signal_noise_ratio(A, An, data_range=1.0)
print(f'Noisy PSNR: {psnr:.2f} dB')

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].imshow(A, cmap='gray', vmin=0, vmax=1)
axes[0].set_title('Original')
axes[0].axis('off')
axes[1].imshow(An, cmap='gray', vmin=0, vmax=1)
axes[1].set_title(f'Noisy (σ={noise_sigma}, PSNR={psnr:.1f}dB)')
axes[1].axis('off')
axes[2].imshow(noise, cmap='bwr', vmin=-3*noise_sigma, vmax=3*noise_sigma)
axes[2].set_title('Noise (赤=正, 青=負)')
axes[2].axis('off')
fig.tight_layout()
plt.show()

sigma_tag = f'{int(noise_sigma*1000):04d}'
out_path = f'data/noisy_{sigma_tag}.pgm'
skimage.io.imsave(out_path, (An * 255).astype(np.uint8))
print(f'{out_path} を保存しました（denoise_bm3d.ipynb で使用）')
```
