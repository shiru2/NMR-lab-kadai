# No.4 DFT手動実装（演習）

**DFT（離散フーリエ変換）の定義式を自分でコードに書きます。**

$$G[k] = \sum_{n=0}^{N-1} g[n] \cdot e^{-j \frac{2\pi kn}{N}}$$

これを実部と虚部に分解すると：

$$\text{Re}[k] = \sum_{n=0}^{N-1} g[n] \cos\left(\frac{2\pi kn}{N}\right)$$

$$\text{Im}[k] = -\sum_{n=0}^{N-1} g[n] \sin\left(\frac{2\pi kn}{N}\right)$$

```python
freq = np.arange(N) * Dx
fig, axes = plt.subplots(2, 1, figsize=(10, 6))
axes[0].plot(freq, re, label='Real')
axes[0].set_title('DFT — Real part')
axes[0].legend()
axes[0].grid(True)
axes[1].plot(freq, im, label='Imaginary', color='orange')
axes[1].set_title('DFT — Imaginary part')
axes[1].legend()
axes[1].grid(True)
fig.tight_layout()
plt.show()
```
