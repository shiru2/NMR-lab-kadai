# No.2 矩形パルス生成と1D FFT

矩形（Rectangular）パルスを生成し、FFTで周波数成分を観察します。

矩形パルスのFTはsinc関数になることを確認してください。

````python
import numpy as np
import matplotlib.pyplot as plt
import japanize_matplotlib
import scipy.io
````

## 矩形パルス生成

`DW` を 17 または 65 に変えて結果の変化を観察してください。

````python
DNUM = 1024
DW = 17   # ← 65 に変えて比較してみよう

Signal = np.zeros(DNUM, dtype=complex)
d1 = DNUM // 2 - DW // 2
d2 = DNUM // 2 + (DW + 1) // 2
Signal[d1:d2] = 1.0

fig, axes = plt.subplots(2, 1, figsize=(10, 5))
axes[0].plot(Signal.real, linewidth=1)
axes[0].set_title(f'Rectangular Pulse (DW={DW}) — Real part')
axes[0].grid(True)
axes[1].plot(Signal.imag, linewidth=1, color='orange')
axes[1].set_title('Imaginary part')
axes[1].grid(True)
fig.tight_layout()
plt.show()

scipy.io.savemat(f'kukei_DW{DW}.mat', {'Signal': Signal})
print(f'kukei_DW{DW}.mat を保存')
````

## FFT

`fftshift(fft(fftshift(x)))` で原点を中心に配置したFFTを行います。

**MATLABとの対応:**
```matlab
Output = fftshift(fft(fftshift(Signal)));
```
```python
Output = np.fft.fftshift(np.fft.fft(np.fft.fftshift(Signal)))
```

````python
flag = 0  # 0: FFT,  1: IFFT

if flag == 0:
    Output = np.fft.fftshift(np.fft.fft(np.fft.fftshift(Signal)))
    title_prefix = 'FFT'
else:
    Output = np.fft.fftshift(np.fft.ifft(np.fft.fftshift(Signal)))
    title_prefix = 'IFFT'

fig, axes = plt.subplots(2, 1, figsize=(10, 5))
axes[0].plot(Output.real, linewidth=1, label='Real')
axes[0].set_title(f'{title_prefix} — Real part (sinc形状)')
axes[0].legend()
axes[0].grid(True)
axes[1].plot(Output.imag, linewidth=1, label='Imaginary', color='orange')
axes[1].set_title(f'{title_prefix} — Imaginary part')
axes[1].legend()
axes[1].grid(True)
fig.tight_layout()
plt.show()

scipy.io.savemat(f'kukei_DW{DW}_FFT.mat', {'Signal': Output})
````
