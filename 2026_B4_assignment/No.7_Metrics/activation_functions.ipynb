# No.7 活性化関数の可視化

ニューラルネットワークで使われる主要な活性化関数をプロットします。

| 関数 | 定義 |
|------|------|
| Step | $f(x) = \begin{cases}1 & x \geq 0 \\ 0 & x < 0\end{cases}$ |
| Sigmoid | $f(x) = \frac{1}{1+e^{-x}}$ |
| ReLU | $f(x) = \max(0, x)$ |

```python
import numpy as np
import matplotlib.pyplot as plt
import japanize_matplotlib
```

```python
# 活性化関数の定義
step    = lambda x: np.where(x >= 0, 1.0, 0.0)
sigmoid = lambda x: 1.0 / (1.0 + np.exp(-x))
relu    = lambda x: np.maximum(0, x)

x = np.linspace(-6, 6, 1200)

fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(x, step(x),    label='Step',    linewidth=3)
ax.plot(x, sigmoid(x), label='Sigmoid', linewidth=3)
ax.plot(x, relu(x),    label='ReLU',    linewidth=3)
ax.axhline(0, color='black', linewidth=0.8)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlim(-6, 6)
ax.set_ylim(-0.5, 2)
ax.set_xlabel('x', fontsize=14)
ax.set_ylabel('y', fontsize=14)
ax.set_title('Activation Functions', fontsize=16)
ax.legend(fontsize=13)
ax.grid(True, alpha=0.4)
fig.tight_layout()
fig.savefig('activation_functions.png', dpi=150)
plt.show()
```
